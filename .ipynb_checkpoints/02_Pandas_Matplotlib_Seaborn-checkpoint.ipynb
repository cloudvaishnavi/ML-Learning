```python
import numpy as np
import pandas as pd
messy_data = {
    "Day": [1, 2, 3, 4, 4, 5, 6],
    "Temperature": [22, 25, np.nan, 32, 32, 35, 38],
    "Power": [450, 600, 850, np.nan, np.nan, 1400, 1600],
    "Appliances": [2, 3, 4, 5, 5, np.nan, 7]
}

messy_df = pd.DataFrame(messy_data)

print(messy_df)
```

```python
messy_df.isnull().sum()
```

```python
messy_df.duplicated().sum()
```

```python
drop_df=messy_df.dropna()
print(drop_df)
```

```python
print(messy_df)
```

```python
filled_df = messy_df.copy()

filled_df["Temperature"] = filled_df["Temperature"].fillna(
    filled_df["Temperature"].mean()
)

print(filled_df)
```

```python
messy_df["Power"].fillna(messy_df["Power"].mean())
```

```python
clean_df = filled_df.drop_duplicates()
```

```python
Q1 = df["Power"].quantile(0.25)
Q3 = df["Power"].quantile(0.75)

print("Q1:", Q1)
print("Q3:", Q3)
IQR = Q3 - Q1

print("IQR:", IQR)
```

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower:", lower)
print("Upper:", upper)
```

```python
outliers = df[
    (df["Power"] < lower) |
    (df["Power"] > upper)
]

print(outliers)
```
