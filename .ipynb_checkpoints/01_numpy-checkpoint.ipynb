```python
import numpy as np

data = np.array([
    [1, 20, 500, 3],
    [2, 25, 750, 4],
    [3, 30, 1000, 5],
    [4, 35, 1250, 6],
    [5, 40, 1500, 7]
])

print(data)
print(data.shape)
```

```python
print(data[:,2].mean())
```

```python
print(data[:, 2].max())
print(data[:, 2].min())
```

```python
high_usage = data[:, 2] > 1000

print(high_usage)
```

```python
high_usage = data[data[:, 2] > 1000]

print(high_usage)
```

```python
power_watts = data[:, 2]

power_kw = power_watts / 1000

print(power_kw)
```

```python
average_power = power_kw.mean()

print(average_power)
```

```python
correction = np.array([0.1, 0.2, 0.1, 0.2, 0.1])
corrected_power = power_kw + correction

print(corrected_power)
```

```python
X = data[:, [1, 3]]

print(X)
print(X.shape)
```

```python
y = data[:, 2]

print(y)
```

```python
weights = np.array([20, 100])
```

```python
prediction = X @ weights

print(prediction)
```

```python
bias = 100

prediction = X @ weights + bias

print(prediction)
```
